```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn
import seaborn as sns
import pandas as pd
from sklearn.metrics import(accuracy_score,precision_score,recall_score,f1_score,confusion_matrix,classification_report)
tokenised_df=pd.read_csv("hugging_tokenised_df.csv")
untokenised_df=pd.read_csv("hugging_untokenised.csv")
validation_df=pd.read_csv('no_duplicate_validation.csv')
```

# FASTTEXT

```python
%pip install gensim
```

```python

from gensim.models import FastText
```

```python
def fasttext(data):
 fasttext_model=FastText(sentences=data,vector_size=100,window=5,min_count=1,workers=4,sg=1)
 x_train_fasttext=np.array([np.mean([fasttext_model.wv[word] for word in sentence],axis=0) if len(sentence)>0 else np.zeros(100) for sentence in data])
 return x_train_fasttext,fasttext_model
```

```python
fasttext_x_train,fastext=fasttext(tokenised_df["text"])

y_train=tokenised_df["label"]
x_val_text=validation_df["text"]
y_val_label=validation_df["label"]
```

```python

 x_val_fasttext = np.array([
    np.mean([fastext.wv[word] for word in sentence], axis=0)
    if len(sentence)>0 else np.zeros(100)
    for sentence in x_val_text])
```

```python
print(fasttext_x_train.shape)
```

```python
fasttext_x_train[0:5]
```

# TRAINING

```python
def train(model):
    return model.fit(fasttext_x_train,y_train)
```

# Metric Function

```python
def Metric(y_val,y_pred):
    accuracy=accuracy_score(y_val,y_pred)
    print("Accuracy=",accuracy)
    precision=precision_score(y_val,y_pred,average="weighted")
    print("precision=",precision)
    recall=recall_score(y_val,y_pred,average="weighted")
    print("recall=",recall)
    f1=f1_score(y_val,y_pred,average="weighted")
    print("F1 score=",f1)
    macro_f1=f1_score(y_val,y_pred,average="macro")
    print("macro F1 Score=",macro_f1)
    cm=confusion_matrix(y_val,y_pred)
    print("Confusion Matrix:")
    print(cm)
    cr=classification_report(y_val,y_pred)
    print("classification Report:")
    print(cr)
```

### LOGISTIC REGRESSION

```python
from sklearn.linear_model import LogisticRegression
logistic_regression_model=LogisticRegression()
train(logistic_regression_model)
print(logistic_regression_model)
```

## Validation

```python
y_pred_lg=logistic_regression_model.predict(x_val_fasttext)
```

```python
print(y_pred_lg)
```

## Metrics

```python
Metric(y_val_label,y_pred_lg)
```

### SVM

```python
from sklearn.svm import SVC
svm_model=SVC()
train(svm_model)
print(svm_model)
```

## validation

```python
print(y_pred_svm)
```

## Metrics

```python
Metric(y_val_label,y_pred_svm)
```
