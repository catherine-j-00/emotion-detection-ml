```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn
import seaborn as sns
import pandas as pd
stop_word_removed_data=pd.read_csv("stop_word_removed_dataset.csv")
```

# Tokenization


```python
train_df=stop_word_removed_data.copy()
```

```python
import spacy
nlp=spacy.load("en_core_web_sm")
```

```python
def tokenization(data):
    data=nlp.pipe(data)
    return [[token.text for token in words]for words in data]
```

# Lemmatization

```python

def lemmatization(data):
    data=[text if isinstance(text,str) else " ".join(text) for text in data]
    data=nlp.pipe(data)
    return [" ".join(token.lemma_ for token in words)for words in data]
```

```python
untokenised_df=train_df.copy()
untokenised_df["text"]=lemmatization(train_df["text"])
```

```python
tokenised_df=untokenised_df.copy()
tokenised_df["text"]=tokenization(untokenised_df["text"])
```
