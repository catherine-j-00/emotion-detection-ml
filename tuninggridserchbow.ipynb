```python
import joblib
from  fun_needed import custom_preprocessor
from  fun_needed import preprocessing
from fun_needed import contradiction_lemmatization
from sklearn.model_selection import GridSearchCV
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
```

```python
train_df=pd.read_csv("train_data.csv")
val_df=pd.read_csv("validation_data.csv")
x_train=train_df['text']
y_train=train_df['label']
untokenised_df=pd.read_csv("hugging_untokenised.csv")
y_val=val_df['label']
x_val=val_df['text']
```

```python
baseline_model=Pipeline([('bow',CountVectorizer(lowercase=True)),('classifier',LogisticRegression(max_iter=1000))])
```

```python
parameter_lg={"bow__ngram_range":[(1,1)],
              "classifier__C":[1.1],
              "classifier__solver":["lbfgs"],
              "classifier__class_weight":["balanced"],
              'classifier__max_iter': [1000],
              'classifier__random_state':[42]
              }
```

```python
gridsearchcv=GridSearchCV(estimator=baseline_model,param_grid=parameter_lg,
                          cv=5,scoring="f1_weighted",n_jobs=1,
                          verbose=2)
```

```python
gridsearchcv.fit(x_train,y_train)
```

```python
print("parameter:\n",gridsearchcv.best_params_)

print("score:\n",gridsearchcv.best_score_)
print("estimator:\n",gridsearchcv.best_estimator_)
```

```python
best_model=gridsearchcv.best_estimator_
joblib.dump(best_model,"baseline_bow_best_gscv_model.pkl")
```
